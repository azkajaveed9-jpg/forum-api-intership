```python
import pandas as pd
import re
```

```python
def clean_my_text(text):
    if not isinstance(text, str):
        return ""
    
    # 1. Links hatana
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    
    # 2. HTML Tags hatana
    text = re.sub(r'<.*?>', '', text)
    
    # 3. Emojis aur baqi faltu symbols hatana
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    
    # 4. Faltu khali jagah khatam karna
    text = " ".join(text.split())
    
    return text
```

```python
# 1. Data load karna
df = pd.read_csv('../forum_posts.csv')

# 2. Machine ko 'Quote' column par chalana (Kyunki aapki file mein yahi naam hai)
df['clean_text'] = df['Quote'].apply(clean_my_text)

# 3. Choti posts hatana (jisme 3 se kam words hon)
df = df[df['clean_text'].str.split().str.len() >= 3]

# 4. Saaf suthri file save karna
df.to_csv('clean_dataset.csv', index=False)

# 5. Data ko screen par dekhna
df.head()
```
